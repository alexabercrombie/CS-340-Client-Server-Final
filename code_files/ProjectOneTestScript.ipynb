```python
from CRUD_Python_Module import AnimalShelter
from bson.objectid import ObjectId 
from pprint import pprint



#Initializing an instance of the CRUD class.
UserSearch = AnimalShelter() 


# This is a custom document that will be added, updated and deleted in the loop below
CujoData  = {
    "name" : "Cujo",
    "age_upon_outcome" : "5 years",
    "breed" : "St Bernard Demon Mix",
    "date_of_birth" : "2021-05-31",
    "outcome_type" : "Transfer",
    "sex_upon_outcome" : "Unknown"
}


# These are the 2 set of data to be updated to all documents named Cujo
UpdateData1 = {
    "$set": {"color" : "Brown and White",
    "sex_upon_outcome" : "Spayed"}
}

UpdateData2 = {
    "$set": {"color" : "Brown",
    "sex_upon_outcome" : "Unknown"}
}


#Search info to find the custom Cujo document added in this .py file
CujoQuery = {"name" : "Cujo", "breed" : "St Bernard Demon Mix"} 


#search to find all Cujo documents
AllCujo = {"name" : "Cujo"} 


# I created a while loop so the user can input the desired actions and vview results.
# I also wanted the actions to be reversable so that it could be repeatable

running = True

while running:
    selection = int(input("Press 1 to add the custom Cujo document.\n"
                          "Press 2 to find and view all documents with a name Cujo.\n"
                          "Press 3 to find and iew the new Cujo document added in option 1.\n"
                          "Press 4 for set 1 of updates.\n"
                          "Press 5 for set 2 of updates\n"
                          "Press 6 to delete the custom Cujo document created in step 1\n"
                          "Press 9 to quit\n"))
    if selection == 1:
        print("Succesfully added a document to aac database: ", UserSearch.create(CujoData), "\n")
    elif selection == 2:
        pprint(UserSearch.read(AllCujo))
    elif selection == 3:
        pprint(UserSearch.read(CujoQuery))
    elif selection == 4:
        print("Documents updated: ", UserSearch.update(AllCujo, UpdateData1))
    elif selection == 5:
        print("Documents updated: ", UserSearch.update(AllCujo, UpdateData2))
    elif selection == 6:
        print("Deleted documents: ", UserSearch.delete(CujoQuery), "\n")
        
    elif selection == 9:
        running = False
        break
    else:
        print("Please enter a valid selection")
```
